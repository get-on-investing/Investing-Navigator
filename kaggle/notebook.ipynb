```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../dataset/investing_benchmarks.csv')
print('Dataset shape:', df.shape)
print(df.head(10))
```

```python
signal_cols = ['portfolio_readiness_score','market_knowledge_score','risk_assessment_score','wealth_building_score','investment_education_score','market_navigation_score']
signal_avgs = df[signal_cols].mean()
colors = ['#10b981','#3b82f6','#f59e0b','#8b5cf6','#ef4444','#06b6d4']
plt.figure(figsize=(12,5))
plt.bar(signal_avgs.index, signal_avgs.values, color=colors)
plt.title('Investing Navigator - Average Signal Scores', fontsize=14, fontweight='bold')
plt.xlabel('Investment Signal')
plt.ylabel('Average Score (0-100)')
plt.xticks(rotation=20)
plt.axhline(y=70, color='green', linestyle='--', alpha=0.5, label='Healthy threshold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
exp_counts = df['experience_level'].value_counts()
plt.figure(figsize=(8,5))
plt.pie(exp_counts, labels=exp_counts.index, colors=colors[:len(exp_counts)], autopct='%1.0f%%')
plt.title('Investing Navigator - Investor Experience Distribution', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
